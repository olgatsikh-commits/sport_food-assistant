## 5. Память между вызовами

```python
config = {'configurable': {'thread_id': 'demo-memory'}}
r1 = graph.invoke({'user_input': 'Я хочу похудеть'}, config=config)
print('1:', r1['answer'])
r2 = graph.invoke({'user_input': 'Что мне съесть на завтрак?'}, config=config)
print('2:', r2['answer'])
state = graph.get_state(config)
print('\nИстория:')
for m in state.values.get('history', []):
    print(f"  [{m['role']}] {m['content'][:100]}...")
```
